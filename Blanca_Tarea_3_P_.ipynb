```python
# Ejercicio 1
p_spam = 0.20
p_no_spam = 0.80

p_oferta_given_spam = 0.80
p_oferta_given_no_spam = 0.10

# P(Oferta) = P(Oferta|Spam)*P(Spam) + P(Oferta|No Spam)*P(No Spam)
p_oferta = (p_oferta_given_spam * p_spam) + (p_oferta_given_no_spam * p_no_spam)

# P(Spam|Oferta) = (P(Oferta|Spam) * P(Spam)) / P(Oferta)
p_spam_given_oferta = (p_oferta_given_spam * p_spam) / p_oferta

print(f"E1: P(Oferta) = {p_oferta:.4f}")
print(f"E1: P(Spam|Oferta) = {p_spam_given_oferta:.4f} ({p_spam_given_oferta*100:.2f}%)")

# Ejercicio 2
p_chrome = 0.40
p_safari = 0.35
p_otros = 0.25

p_compra_given_chrome = 0.03
p_compra_given_safari = 0.04
p_compra_given_otros = 0.02

# P(Compra)
p_compra = (p_chrome * p_compra_given_chrome) + (p_safari * p_compra_given_safari) + (p_otros * p_compra_given_otros)

p_chrome_given_compra = (p_chrome * p_compra_given_chrome) / p_compra
p_safari_given_compra = (p_safari * p_compra_given_safari) / p_compra
p_otros_given_compra = (p_otros * p_compra_given_otros) / p_compra

print(f"E2: P(Compra) = {p_compra:.4f}")
print(f"E2: P(Chrome|Compra) = {p_chrome_given_compra:.4f} ({p_chrome_given_compra*100:.2f}%)")
print(f"E2: P(Safari|Compra) = {p_safari_given_compra:.4f} ({p_safari_given_compra*100:.2f}%)")
print(f"E2: P(Otros|Compra) = {p_otros_given_compra:.4f} ({p_otros_given_compra*100:.2f}%)")
```
